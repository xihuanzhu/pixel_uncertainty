```python
import numpy as np
import  os
```

```python
def npy_to_single_npy(npy_data, save_dir, index):
    if not os.path.exists(save_dir):
        os.makedirs(save_dir)
    for i,k in enumerate(npy_data):
        np.save(save_dir + str(i + index) + '.npy', k)
        #print(i, "&&&&&", k)
```

```python
data_path = "/raid/zhuxihuan/data/body/test2_npy_data/"
single_data_save_path = "/raid/zhuxihuan/data/body/MC_uncertainty_single_data/"
train_slice_save_dir = single_data_save_path + "train_slice/"
train_mask_save_dir =  single_data_save_path + "train_mask/"
if not os.path.exists(single_data_save_path):
    os.makedirs(single_data_save_path)
if not os.path.exists(train_slice_save_dir):
    os.makedirs(train_slice_save_dir)
if not os.path.exists(train_mask_save_dir):
    os.makedirs(train_mask_save_dir)
num = 0
for sdir in os.listdir(data_path):
    print(sdir)
    mask_path = os.path.join(data_path, sdir) + "/mask.npy"
    print(mask_path)
    train_mask = np.load(mask_path)
    slice_path = os.path.join(data_path, sdir) + "/slice.npy"
    train_slice = np.load(slice_path).astype(np.float64)
    print(train_slice.shape[0])
    #print(np.min(train_slice), "*****", np.max(train_slice), "*****", np.min(train_mask), "*****", np.max(train_mask), "*****")
    npy_to_single_npy(train_slice, train_slice_save_dir, num)
    npy_to_single_npy(train_mask, train_mask_save_dir, num)
    num += train_slice.shape[0]
[58, 24, 59, 30, 89, 
 54, 21, 1, 19, 29, 
 79, 15, 11, 26, 28,
 38, 3, 27, 91, 33, 
 75, 48, 44, 90, 82,
 43, 20, 7, 67, 42,
 80, 36, 53, 84, 72,
 10, 51, 25, 17, 60,
 73, 86, 83, 23, 62, 
 68, 50, 22]
```
