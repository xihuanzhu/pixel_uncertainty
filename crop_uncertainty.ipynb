```python
gt_path = "/raid/zhuxihuan/data/body/expe4/val_data/GT"
prob = "/raid/zhuxihuan/data/body/expe4/val_data/oprobs"
for s in range(len(os.listdir(gt_path)[:])):
    gt = np.load(gt_path + f"/{s}.npy")
    probs = np.load(prob + f"/{s}.npy")
    gt[gt == 1] = 255
    probs = probs[:, :, -1]
    probs[probs >= 0.5] = 255
    probs[probs < 0.5] = 0
    print(gt.shape, probs.shape)
    f_img = np.concatenate( (gt,probs), axis=1 )
    image = Image.fromarray(f_img.astype('uint8'))
    image.save(f"./kk/img{s}.jpg")
    #plt.imshow(probs)
```
