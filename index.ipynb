```python
def sigmoid(x):
    s = 1 / (1 + np.exp(-x))
    return s
def myvar(data):
    #data = sigmoid(data)
    mean = np.mean(data)
    var = 0
    for i in range(len(data)):
        var = var + (data[i] - mean) * (data[i] - mean)
    #var = np.var(data)
    return var/len(data)
data = np.array([0.5, 0.5, 0.5, 1.0, 1.0])
print(myvar(data))
# 0.005600000000000004 0.007
# 0.06    0.075
# 0.16    0.2
```

```python
def entropy(data):
    mean = np.mean(data)
    return - mean * np.log(mean) - (1 - mean) * np.log(1 - mean)
print(entropy(data))
```

```python
def MI(data):
    mean = np.mean(data)
    g = np.mean(-data * np.log(data + np.finfo(np.float32).eps) - (1 - data) * np.log(1 - data + np.finfo(np.float32).eps))
    return - mean * np.log(mean) - (1 - mean) * np.log(1 - mean) - g
print(MI(data))
```
