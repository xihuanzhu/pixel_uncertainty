```python
import numpy as np 
from PIL import Image
import os 
```

```python
data_dir = "/raid/zhuxihuan/data/body/test_npy_data/"
data_visual_save_dir = "/raid/zhuxihuan/data/body/data_visual"
if not os.path.exists(data_visual_save_dir):
    os.makedirs(data_visual_save_dir)
for sdir in os.listdir(data_dir):
    print(sdir)
    save_path = os.path.join(data_visual_save_dir, sdir)
    if not os.path.exists(save_path):
        os.makedirs(save_path)
    print(save_path)
    slice_path = os.path.join(data_dir, sdir) + "/slice.npy"
    mask_path = os.path.join(data_dir, sdir) + "/mask.npy"
    sslice = np.load(slice_path)
    sslice = np.clip(sslice, -1000, 1000)
    mmask = np.load(mask_path)
    print(sslice.shape, mmask.shape)
    for i in range(sslice.shape[0]):
        s = np.squeeze(sslice[i])
        m = np.squeeze(mmask[i]) * 255
        image = Image.fromarray(s.astype('uint8'))
        image.save(save_path + f"/slice{i}.png")
        image = Image.fromarray(m.astype('uint8'))
        image.save(save_path + f"/mask{i}.png")
```
